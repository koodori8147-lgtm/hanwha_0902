```python
#logprob 예제

from openai import OpenAI
import math

client = OpenAI()

question = "스타벅스 아이스 아메리카노 그란데 사이즈의 가격은 얼마인가? 가격만 숫자와 원 단위로 답하세요."

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {"role": "user", "content": question}
    ],
    logprobs=True,
    top_logprobs=5
)

print("답변")
print(response.choices[0].message.content)

print("\n=== 토큰별 확률 정보 ===")
for token_info in response.choices[0].logprobs.content:

    token = token_info.token
    logprob = token_info.logprob

    probability = math.exp(logprob)

    print(f"Token: {repr(token):15}"
          f" Logprob: {logprob:10.5f}"
          f" 확률: {probability * 100:8.3f}%")

tokens = response.choices[0].logprobs.content

response = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "user",
            "content": question
        }
    ],
    temperature=0.9,
    logprobs=True,
    top_logprobs=5
)

for token_info in response.choices[0].logprobs.content:

    print("\n선택된 토큰:", repr(token_info.token))
    print(
        "선택 확률:",
        round(math.exp(token_info.logprob) * 100, 2),
        "%"
    )

    print("다른 후보:")

    for candidate in token_info.top_logprobs:
        print(
            repr(candidate.token),
            round(math.exp(candidate.logprob) * 100, 2),
            "%"
        )
```
