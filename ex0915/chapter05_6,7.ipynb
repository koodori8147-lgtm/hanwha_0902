```python
from langchain_core.prompts.few_shot import FewShotPromptTemplate
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_teddynote import logging
from dotenv import load_dotenv

logging.langsmith("test_0915")
load_dotenv()

examples=[
    {
        "question" : "서울의 음식점인 하쿠텐과 쿠이도라쿠 중에 어떤 곳이 더 맛있나요?",
        "answer": """이 질문에 추가 질문이 필요한가요: 예.
                추가 질문 : 하쿠텐의 구글맵, 카카오맵 평점은 얼마인가요?
                중간 답변 : 하쿠텐의 구글맵 평점은 4.5, 카카오맵 평점은 4.7입니다.
                추가 질문 : 쿠이도라쿠의 구글맵, 카카오맵 평점은 몇 점인가요?
                중간 답변 : 쿠이도라쿠의 구글맵 평점은 4.7, 카카오맵 평점은 4.4입니다.
                최종 답변은 : 하쿠텐
                """,
    },
]

example_prompt = PromptTemplate.from_template(
    "Question:\n{question}\nAnswer:\n{answer}"
)

print(example_prompt.format(**examples[0]))
```

```python
prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="Question:\n{question}\nAnswer:",
    input_variables=["question"],
)

question="대한민국 최고의 라멘집은 어디인가요?"
final_prompt = prompt.format(question=question)
print(final_prompt)
```

```python
from langchain_openai import ChatOpenAI
from langchain_teddynote.messages import stream_response

llm = ChatOpenAI()

answer = llm.stream(final_prompt)
stream_response(answer)
```

```python
prompt = FewShotPromptTemplate(
    examples=examples,
    example_prompt=example_prompt,
    suffix="Question:\n{question}\nAnswer:",
    input_variables=["question"],
)

chain = prompt | llm | StrOutputParser()

answer= chain.stream(
    {"question": "대한민국 라멘의 최고 맛집은 어디인가요?"}
)

stream_response(answer)
```
