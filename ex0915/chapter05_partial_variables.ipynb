```python
template = "{city1}과 {city2}의 라엔 맛집은 각각 어디인가요?"

prompt = PromptTemplate(
    template= template,
    input_variables=["city1"],
    partial_variables={
        "city2": "오사카"
    },
)

prompt
```

```python
prompt.format(city1="서울")
```

```python
prompt_partial=prompt.partial(city2="도쿄")
prompt_partial
```

```python
prompt_partial.format(city1="대전")
```

```python
chain = prompt_partial | llm

chain.invoke("서울").content
```

```python
chain.invoke({"city1": "대전", "city2": "옌청"}).content
```

```python
from datetime import datetime

datetime.now().strftime("%B %d")
```

```python
def get_today():
    return datetime.now().strftime("%B %d")

prompt = PromptTemplate(
    template = "Today date is {today}. Please tell me the {n} of Influencers who is birthday at today. And also, please type their birth year and date.",
    input_variables=["n"],
    partial_variables={
        "today": get_today
    },
)

prompt.format(n=5)
chain = prompt | llm
print(chain.invoke(5).content)
```

```python
print(chain.invoke({"today": "jan 02", "n" : 5}).content)
```
