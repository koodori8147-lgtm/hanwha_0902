```python
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain.agents import create_agent

# 1. Tool 정의
@tool
def get_famous_restaurant(city: str) -> str:
    """Get a famous restaurant for a given city."""
    restaurants = {
        "서울": "명동교자",
        "부산": "해운대암소갈비집",
        "대전": "성심당",
        "대구": "왕거미식당",
        "광주": "송정떡갈비"
    }

    return  restaurants.get(city,f"{city}의 유명한 식당 정보를 찾을 수 없습니다.")

# 2. 모델 및 에이전트 연결
llm = ChatOpenAI(model="gpt-4o-mini", temperature=1.0)
agent = create_agent(model=llm, tools=[get_famous_restaurant])

# 3. 실행
result = agent.invoke({"messages": [("user", "대전의 유명한 식당 알려줘")]})
print(result["messages"][-1].content)

# 랭스미스 Tracing에서 확인하기
```
