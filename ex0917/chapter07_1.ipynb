```python
%%time
from dotenv import load_dotenv
from langchain_teddynote import logging

load_dotenv()
logging.langsmith("test_0917")

from langchain_openai import ChatOpenAI
from langchain_core.prompts import PromptTemplate

llm = ChatOpenAI(model_name="gpt-4o", temperature = 0.1)

prompt = PromptTemplate.from_template(
    """
    입력된 도시: {city}

    반드시 입력된 도시 안에 위치한 라멘 음식점만 추천해줘.
    다른 도시나 다른 국가의 음식점은 절대 포함하지 마.
    
    {city}의 라멘 맛집 Top 5를 선정하고,
    각 음식점의 대표 메뉴를 간단히 요약해줘.

    음식점 위치가 {city}인지 확신할 수 없다면 추천하지 마.
    """
)

chain = prompt | llm

response = chain.invoke({"city": "서울"})
print(response.content)
```

```python
%%time
from langchain_core.globals import set_llm_cache
from langchain_core.caches import InMemoryCache

set_llm_cache(InMemoryCache())

response = chain.invoke({"city": "나고야"})
print(response.content)
```
